```python
# create a simple tokenizer class
class SimpleTokenizer:

    def __init__(self , vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for i,s in enumerate(vocab)}

    # encoder (string--> integers list)
    def encoder(self,text_sample):
        # preprocess
        result = re.split(r'([,.:;?_!"()\']|--|\s)', text_sample)
        preprocessed_sample = [item for item in result if item.strip()]

        #convert to the token ids taken from the vocab
        tokens = [self.str_to_int[i] for i in preprocessed_sample]

        return tokens

    # decoder (int list --> original string)
    def decoder(self , tokens_list):
        # here we need to just find the acctual word
        word = " ".join(self.int_to_str[i] for i in tokens_list)
        # replace the puctuations
        word = re.sub(r'\s+([,.?!"()\'])', r'\1', word)

        return word
```

```python
# create a simple tokenizer class
class SimpleTokenizerV2:

    def __init__(self , vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for i,s in enumerate(vocab)}

    # encoder (string--> integers list)
    def encoder(self,text_sample):
        # preprocess
        result = re.split(r'([,.:;?_!"()\']|--|\s)', text_sample)
        preprocessed_sample = [item for item in result if item.strip()]

        # add the new UNK token
        preprocessed_sample = [
            item if item in self.str_to_int
            else "<|unk|>" for item in preprocessed_sample

        ]
        #convert to the token ids taken from the vocab
        tokens = [self.str_to_int[i] for i in preprocessed_sample]

        return tokens

    # decoder (int list --> original string)
    def decoder(self , tokens_list):
        # here we need to just find the acctual word
        word = " ".join(self.int_to_str[i] for i in tokens_list)
        # replace the puctuations
        word = re.sub(r'\s+([,.?!"()\'])', r'\1', word)

        return word
```

```python
from torch.utils.data import Dataset , DataLoader

class GPTDatasetV1(Dataset):
    def __init__(self , txt , tokenizer , max_length , stride):
        self.input_ids = []
        self.target_ids = []

        # first tokenize the text
        token_ids = tokenizer.encode( txt ,allowed_special={'<|endoftext|>'} )

        # use the sliding window
        for i in range(0 ,len(token_ids) - max_length , stride):
            
            input_chunks = token_ids[i : i + max_length]
            target_chunks = token_ids[i+1 : i + max_length+1]

            # convert to tensors
            self.input_ids.append(torch.tensor(input_chunks))
            self.target_ids.append(torch.tensor(target_chunks))

    
    def __len__(self):
        return len(self.input_ids)
    
    def __getitem__(self , index):
        return self.input_ids[index] , self.target_ids[index]
```

```python
def create_dataloader_v1(txt ,batch_size=4, max_length=256, 
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):
    
    # Initialize the tokenizer
    tokenizer = tiktoken.get_encoding("gpt2")

    # Create dataset
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

    # create the dataloader
    dataloader =  DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        drop_last=drop_last,
        num_workers=num_workers

    )


    return dataloader
```

```python
# Let's test the dataloader with a batch size of 1 for an LLM with a context size of 4:
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
```
